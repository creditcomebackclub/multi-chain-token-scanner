# Data audit

This notebook checks sample composition, missingness, repeated token/support groups, and label rates before any model is fit. Every candidate feature must be derivable from closed candles at or before `detected_at`; the reviewed provenance lives in `scanner_ml.data.LEAKAGE_REVIEW`.

Synthetic data may exercise this notebook, but it is never evidence of a trading edge.

```python
from pathlib import Path
import sys
from IPython.display import display
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path("research/ml")
sys.path.insert(0, str(ROOT))
SNAPSHOT = ROOT / "data" / "snapshot.csv"
from scanner_ml.data import FEATURES, LEAKAGE_REVIEW, audit, load_dataset
if not SNAPSHOT.exists():
    print("Awaiting production snapshot; run npm run research:export")
else:
    data = load_dataset(SNAPSHOT)
    summary = audit(data)
    display({key: summary[key] for key in ("rows", "resolved", "by_source", "by_chain", "by_label", "groups", "repeated_groups")})
    display(data[FEATURES].describe().T.assign(missing_pct=data[FEATURES].isna().mean() * 100))
    display(LEAKAGE_REVIEW)
```
